```python
# url
youtube_link="https://www.youtube.com/watch?v=M_5oEvAw_Og"  # youtub video id = M_5oEvAw_Og
```

```python
# Import YouTubeTranscriptApi for fetching video transcripts
from youtube_transcript_api import YouTubeTranscriptApi

youtube_id = "M_5oEvAw_Og"

try:
    ytt_api = YouTubeTranscriptApi()

    # Fetch transcript with Hindi language support
    transcript = ytt_api.fetch(
        youtube_id,
        languages=['hi']
    )
    transcript_data = transcript.to_raw_data()

    # Print each text line from the transcript
    for item in transcript_data:
        print(item['text'])

except Exception as e:
    print(f"Error fetching transcript: {e}")
```
